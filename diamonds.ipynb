```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=sns.load_dataset('diamonds')
df.head()
```

```python
print(df)
```

```python
regression=df[["carat","price"]].dropna()
regression=regression[regression["carat"]<=3]
print(regression)
```

```python
x=regression["carat"].to_numpy()
y=regression["price"].to_numpy()
```

```python
x_mean=np.mean(x)
y_mean=np.mean(y)
covariance=np.sum((x-x_mean)*(y-y_mean))
d=np.sum((x-x_mean)**2)
slope=covariance/d
intercept=y_mean-slope*x_mean
y_predicted=slope*x+intercept
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x,y,color="skyblue",alpha=0.7,label="Actual data")
plt.plot(x,y_predicted,label="Regression data",color="red",alpha=1)
plt.legend()
plt.show()
```

```python
#storing a data_Set column as a numpy array
clean_data=df.dropna(subset=["total_bill","tip"])
x=clean_data["total_bill"].to_numpy()
y=clean_data["tip"].to_numpy()
# print(x)
# print(y)
```

```python
# plt.figure(figsize=(10,6))
# sns.scatterplot(clean_data,x="total_bill",y="tip")
# plt.show()
```

```python
#calculationg slope of regression equation
x_mean=np.mean(x)
y_mean=np.mean(y)
cov=np.sum((x-x_mean)*(y-y_mean))
d=np.sum((x-x_mean)*(x-x_mean))
print(cov)
print(d)
slope=cov/d
print(slope)
```

```python
#calculating intercept
b=y_mean - slope*x_mean
print(b)
```

```python
y_p=slope*x+b
fig = go.Figure()
fig.add_trace(go.Scatter(x=x,y=y,mode='markers',name='Actual data',marker=dict(color='skyblue',opacity=0.7,size=8)))
fig.add_trace(go.Scatter(x=x,y=y_p,mode='lines',name='Regression line',line=dict(color='red', width=3)))
fig.update_layout(title="Linear regression on 'tips' data set",xaxis_title="Total Bill",yaxis_title="Tip Amount",template="plotly_white",width=900, height=600)
fig.show()
```

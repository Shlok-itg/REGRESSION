```python
#storing a data_Set column as a numpy array
clean_data=df.dropna(subset=["total_bill","tip"])
x=clean_data["total_bill"].to_numpy()
y=clean_data["tip"].to_numpy()
# print(x)
# print(y)
```

```python
# plt.figure(figsize=(10,6))
# sns.scatterplot(clean_data,x="total_bill",y="tip")
# plt.show()
```

```python
#calculationg slope of regression equation
x_mean=np.mean(x)
y_mean=np.mean(y)
cov=np.sum((x-x_mean)*(y-y_mean))
d=np.sum((x-x_mean)*(x-x_mean))
print(cov)
print(d)
slope=cov/d
print(slope)
```

```python
#calculating intercept
b=y_mean - slope*x_mean
print(b)
```

```python
y_p=slope*x+b
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='skyblue',alpha=0.5,label="actual data")
plt.plot(x,y_p,color='red',alpha=1,linewidth=2,label="regression data")
plt.title("Linear regression on 'tips' data set")
plt.xlabel("total bill")
plt.ylabel("tip amount")
plt.legend()
plt.show()
```

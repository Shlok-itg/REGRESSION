```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=sns.load_dataset("mpg")
df.head()
```

```python
df.loc[:,["mpg","weight"]]
```

```python
regression=df[["mpg","horsepower"]].dropna()
print(regression)
```

```python
x=regression["horsepower"].to_numpy()
y=regression["mpg"].to_numpy()
```

```python
x_mean=np.mean(x)
y_mean=np.mean(y)
cov=np.sum((x-x_mean)*(y-y_mean))
d=np.sum((x-x_mean)**2)
slope=cov/d
intercept=y_mean-slope*x_mean
y_predicted=slope*x+intercept
```

```python
plt.figure(figsize=(10,6))
plt.scatter(x,y,color="skyblue",alpha=0.6,label="actual data")
plt.plot(x,y_predicted,alpha=1,color="red",label="regression_data")
plt.xlabel("Horsepower")
plt.ylabel("mpg")
plt.legend()
plt.plot()
```
